```python
import re
import nltk
from nltk.corpus import stopwords
from gensim.models import Word2Vec
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
text = """
Artificial Intelligence is one of the most important technologies in the modern world.
It enables computers to learn from data, solve problems, and make decisions.
Machine learning is a branch of artificial intelligence that allows systems to improve
their performance through experience. AI is widely used in healthcare, education,
banking, transportation, and many other fields. As technology continues to advance,
artificial intelligence is becoming an essential part of our daily lives.
"""

print(text[:500])
```

```python
# Convert to lowercase
text = text.lower()

# Remove digits
text = re.sub(r'\d+', '', text)

# Remove special characters and punctuation
text = re.sub(r'[^a-z\s]', ' ', text)

# Remove extra spaces
text = re.sub(r'\s+', ' ', text)

print(text[:500])
```

```python
sentences = nltk.sent_tokenize(text)

tokenized_sentences = []

for sentence in sentences:
    words = nltk.word_tokenize(sentence)
    tokenized_sentences.append(words)

print(tokenized_sentences[:2])
```

```python
stop_words = set(stopwords.words('english'))

cleaned_sentences = []

for sentence in tokenized_sentences:
    words = [word for word in sentence if word not in stop_words]
    if words:
        cleaned_sentences.append(words)

print(cleaned_sentences[:3])
```

```python
model = Word2Vec(
    sentences=cleaned_sentences,
    vector_size=100,
    window=2,
    min_count=2,
    sg=0,
    workers=4,
    epochs=10
)

print("CBOW model trained successfully.")
```

```python
words = list(model.wv.index_to_key)

print("Vocabulary size:", len(words))
print("First 20 words:", words[:20])
```

```python
word = "intelligence"

if word in model.wv:
    print(model.wv[word])
else:
    print("Word not found in vocabulary.")
```

```python
word = "intelligence"

if word in model.wv:
    similar_words = model.wv.most_similar(word, topn=10)
    print(similar_words)
```

```python
word1 = "intelligence"
word2 = "artificial"

if word1 in model.wv and word2 in model.wv:
    similarity = model.wv.similarity(word1, word2)
    print("Similarity:", similarity)
```

```python
import numpy as np

context_words = ["artificial", "intelligence"]

vectors = [
    model.wv[word]
    for word in context_words
    if word in model.wv
]

if vectors:
    context_vector = np.mean(vectors, axis=0)

    result = model.wv.similar_by_vector(
        context_vector,
        topn=10
    )

    print("Predicted words:")
    for word, score in result:
        print(word, score)
```
